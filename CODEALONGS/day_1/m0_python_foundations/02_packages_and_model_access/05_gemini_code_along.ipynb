# Code Along: Call A Hosted Model

Start with one Gemini request. Then make the prompt and model choice visible.

## 1. Make one call

Set `GEMINI_API_KEY` first. Which line sends data outside your laptop?

```python
import os

from dotenv import load_dotenv
from google import genai

load_dotenv(override=True)
client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
response = client.models.generate_content(
    model="gemini-2.5-flash-lite", contents="Say hello."
)
print(response.text)
```

## 2. Your turn

Change only `contents` to ask for a one-sentence portfolio observation. What is the instruction? What is the data?

```python
instruction = "Answer in one short sentence."
data = "Alice holds 52% of her portfolio in AAPL."
response = client.models.generate_content(
    model="gemini-2.5-flash-lite",
    contents=f"{instruction}\n{data}",
)
print(response.text)
```
