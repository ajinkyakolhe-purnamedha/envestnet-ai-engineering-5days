# FastAPI server

A route decorator connects a URL to a Python function.

```python
from fastapi import FastAPI


app = FastAPI()


@app.get("/health")
def health() -> dict[str, str]:
    """Confirm that the application is running."""
    return {"status": "ok"}
```

In a terminal run `uv run uvicorn wealth_demo.server:app --reload`, then call `curl http://localhost:8000/health` or open `/docs`.

```python
@app.get("/portfolio")
def portfolio() -> dict:
    """Return the synthetic portfolio as JSON."""
    return {"symbol": "AAPL", "shares": 10}
```
