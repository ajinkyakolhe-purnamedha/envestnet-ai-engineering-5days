## 1. Buy ten shares

Before you run this: how much will Alice spend, and how much of $100,000 remains?

```python
cash = 100_000
shares = 10
historical_price = 80.50

cost = shares * historical_price
cash_left = cash - cost

print("Cost:", cost)
print("Cash left:", cash_left)
```

## 3. What is it worth today?

The historical price tells us what Alice paid. A latest price answers a different question.

```python
latest_price = 120.00
current_value = shares * latest_price
gain = current_value - cost

print("Value today:", current_value)
print("Gain:", gain)
```

## 4. Add MSFT

Write the second purchase with ordinary variables first. The repetition is deliberate.

```python
msft_shares = 5
msft_historical_price = 182.83
msft_cost = msft_shares * msft_historical_price

total_paid = cost + msft_cost
cash_left = cash - total_paid
print("Total paid:", total_paid)
print("Cash left:", cash_left)
```

## 5. Group repeated facts

A dictionary keeps the facts for one purchase together; a list holds several purchases.

```python
purchases = [
    {"symbol": "AAPL", "shares": 10, "price": 80.46},
    {"symbol": "MSFT", "shares": 5, "price": 182.83},
]

for purchase in purchases:
    cost = purchase["shares"] * purchase["price"]
    print(purchase["symbol"], cost)
```

## 6. Name the repeated calculation

Now that the same multiplication appears twice, a function earns its place.

```python
def purchase_cost(shares, price):
    return shares * price

total_paid = sum(
    purchase_cost(item["shares"], item["price"])
    for item in purchases
)
print("Total paid:", total_paid)
```
