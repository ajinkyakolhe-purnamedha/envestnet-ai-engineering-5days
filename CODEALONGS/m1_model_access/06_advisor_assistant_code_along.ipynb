# Code-along · The application boundary

The snippet builds a grounded prompt, measures both sides of the boundary, and
takes the model as a parameter. That last choice is what this notebook is about.


```python
from runpy import run_path

snippet = run_path("05_advisor_assistant.py")
```

## Explore

1. Below, a fake model replaces the real one. No key, no network, no cost —
   and the assistant still works. This is how every later module tests.
2. Print the prompt the fake model received. Is every number in it a fact you
   supplied, or did the code invent one?
3. Change a value in `FACTS` and re-run. The prompt changes; the code does not.
4. Now make `fake_model` return `""`. What should a client-facing assistant do
   with no usable text? That is the lab.


```python
answer = snippet["answer"]
facts = dict(snippet["FACTS"])

seen = []

def fake_model(prompt: str) -> str:
    """A model stub: records its input, returns a fixed reply."""
    seen.append(prompt)
    return "AAPL is 52% of invested assets, the largest single holding."

print(answer("How concentrated is this portfolio?", facts, fake_model))

print("\n--- the prompt the model actually received ---")
print(seen[0])
```
