# Python Portfolio
Build a small virtual portfolio from basic Python values.

## 1. Start with values

```python
cash = 25_000
symbol = "SPY"
shares = 100
price = 228.80
print(cash, symbol, shares, price)
```

## 2. Calculate one position

```python
position_value = shares * price
print(position_value)
```

## 3. Describe it

```python
sentence = f"{shares} shares of {symbol} are worth ${position_value}"
print(sentence)
```

## 4. Use a dictionary

```python
holding = {"symbol": symbol, "shares": shares, "price": price}
print(holding)
```

## 5. Store three holdings

```python
holdings = [
    holding,
    {"symbol": "QQQ", "shares": 150, "price": 170.70},
    {"symbol": "GLD", "shares": 75, "price": 140.11},
]
print(holdings)
```

## 6. Loop through them

```python
for item in holdings:
    print(item["symbol"], item["shares"] * item["price"])
```

## 7. Extract a function

```python
def calculate_value(shares, price):
    return shares * price

print(calculate_value(10, 250))
```

## 8. Add type hints

```python
def typed_value(shares: int, price: float) -> float:
    return shares * price

invested_value = sum(typed_value(item["shares"], item["price"]) for item in holdings)
print(invested_value)
```

## 9. Include cash

```python
total_value = cash + invested_value
print(total_value)
```

## 10. Deliberate failure: wrong key

```python
try:
    print(holding["ticker"])
except KeyError as error:
    print("Missing key:", error)
```

## 11. Correct the key access

```python
print(holding["symbol"])
```

## 12. Final result

```python
print(f"Portfolio: ${invested_value:.2f} invested + ${cash:.2f} cash = ${total_value:.2f}")
```

## Takeaways
- Variables hold portfolio facts.
- Collections and loops handle several holdings.
- A function keeps one calculation consistent.
