## Explore

1. Open the model folder below. Which file holds the weights, and which holds
   the vocabulary?
2. The weights are split across four shards. Why would a repository do that?
3. `HF_HUB_OFFLINE=1` makes a missing file fail loudly. What would happen in a
   room of thirty people without it?
4. Compare `num_hidden_layers` and `hidden_size` here against a frontier model.
   M2 explains why that gap shows up in the answers.


```python
import json

model_dir = snippet["MODEL_PATH"]

for path in sorted(model_dir.iterdir()):
    print(f"{path.stat().st_size / 1e6:8.1f} MB  {path.name}")

config = json.loads((model_dir / "config.json").read_text())
for key in ("num_hidden_layers", "hidden_size", "vocab_size"):
    print(f"\n{key:<20} {config[key]:,}")
```
